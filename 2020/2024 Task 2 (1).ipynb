```python
#Task 2.1

class Node:
    def __init__(self):
        self.data_value = -1
        self.left_node = -1
        self.right_node = -1
        
    def set_data(self, value):
        self.data_value = value
    
    def set_left(self, value):
        self.left_node = value
        
    def set_right(self, value):
        self.right_node = value
        
    def get_data(self):
        return self.data_value
    
    def get_left(self):
        return self.left_node
    
    def get _right(self):
        return self.right_node
    
```

```python
#Task 2.2

def calculate_nodes(levels):
        total = -1
        for i in range(levels):
            total = 2 ** (levels) -1
        return f'A { levels} level number tree has {total} nodes'
    
```
