```python
# Task 1.1

import sqlite3

conn = sqlite3.connect("TESTDATABASE.db")

conn.execute("DROP TABLE IF EXISTS Student_Test")
conn.execute("DROP TABLE IF EXISTS Student")
conn.execute("DROP TABLE IF EXISTS Test")
conn.execute("DROP TABLE IF EXISTS Class")

conn.execute("""
CREATE TABLE Student(
    StudentID INTEGER PRIMARY KEY,
    GivenName TEXT,
    FamilyName TEXT,
    ClassName TEXT,
    FOREIGN KEY (ClassName) REFERENCES Class(ClassName))
    """)

conn.execute("""
CREATE TABLE Test(
    TestID TEXT PRIMARY KEY,
    MaxMarks INTEGER)
    """)

conn.execute("""
CREATE TABLE Class(
    ClassName TEXT PRIMARY KEY,
    YearGroup INTEGER)
    """)

conn.execute("""
CREATE TABLE Student_Test(
    StudentTestID INTEGER PRIMARY KEY,
    StudentID INTEGER,
    TestID TEXT,
    Mark INTEGER,
    DateTest TEXT,
    FOREIGN KEY (StudentID) REFERENCES Student(StudentID),
    FOREIGN KEY (TestID) REFERENCES Test(TestID))
    """)

conn.commit()

cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table'")
print(cursor.fetchall())

conn.close()
```
