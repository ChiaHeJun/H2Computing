```python
#Task 4.1
import sqlite3
conn = sqlite3.connect('library.db')
conn.execute(''' DROP TABLE IF EXISTS Book,
    DROP TABLE IF EXISTS Member,
    DROP TABLE IF EXISTS Loan,
    ''')
conn.execute('''CREATE TABLE Book(
    BookID INTEGER PRIMARY KEY, 
    Title TEXT,
    Genre TEXT)
    ''')
conn.execute('''CREATE TABLE Member(
    MemberNumber INTEGER PRIMARY KEY, 
    FamilyName TEXT,
    GivenName TEXT)
    ''')
conn.execute('''CREATE TABLE Loan(
    LoanID INTEGER PRIMARY KEY, 
    MemberID INTEGER,
    BookID INTEGER,
    DateLoaned TEXT,
    Returned TEXT,
    FOREIGN KEY (MemberNumber) REFERENCES Member(MemberNumber),
    FOREIGN KEY (BookID) REFERENCES Book(BOOKID))
    ''')
conn.commit()
conn.close()
```

```python
#Task 4.2
with open('BOOK.txt', 'r') as file: #file closes automatically
    for line in file:
        book_id, title, genre = line.strip().split(',')
        conn.execute('''INSERT INTO Book(
        BookID, Title, Genre) VALUES(?,?,?)'''
        ,(book_id, title, genre))
        
conn.commit()

recs = conn.execute('SELECT * FROM Book')
for rec in recs:
    print (rec)

with open('MEMBER.txt', 'r') as file: #file closes automatically
    for line in file:
        member_number, family_name, given_name = line.strip().split(',')
        conn.execute('''INSERT INTO Member(
        member_number, family_name, given_name) VALUES(?,?,?)'''
        ,(member_number, family_name, given_name))

with open('LOAN.txt', 'r') as file: #file closes automatically
    for line in file:
        loan_id, member_number, book_id, loan_date, returned = line.strip().split(',')
        conn.execute('''INSERT INTO Loan(
        loan_id, member_number, book_id, loan_date, returned) VALUES(?,?,?,?,?)'''
        ,(loan_id, member_number, book_id, loan_date, returned))

conn.commit()
conn.close()
```
