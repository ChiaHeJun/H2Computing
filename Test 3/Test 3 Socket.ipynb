```python
#server

import socket
server = socket.socket()
server.bind(('127.0.0.1', 12345))
server.listen()

print('Waiting for client connection...')

conn, addr = server.accept()

data = conn.recv(1024).decode()

parcelID, time, message = data.strip().split('|')
print('----UPDATE for Parcel Received----')
print('Parcel ID:', parcelID)
print('Time:', time)
print('Message:', message)
print("----------------------------------")

conn.send("Received\n".encode())

conn.close()
server.close()
```

```python
#client

import socket
client = socket.socket()
client.connect(('127.0.0.1', 12345))

parcelID = input('Enter Parcel ID:')
time = input('Enter Time:')
message = input('Enter Status Message:')

data = f"{parcelID}|{time}|{message}\n"
client.sendall(data.encode())

client.close()
```
