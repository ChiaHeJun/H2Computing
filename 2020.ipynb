```python
#Task 1.1
def task1_1(string_value):
    h = 0

    for i in range(len(string_value)):
        val = 33 * ord(string_value[i])
        h = (h + val) % 1024

    return h

hello = task1_1('Hello')
hallo = task1_1('Hallo')
hullo = task1_1('Hullo')

print(hello)
print(hallo)
print(hullo)

#Task 1.2
def task1_2(seed, string_value):
    combined = seed + string_value
    return task1_1(combined)

print(task1_2("seed-one", "Hello"))
print(task1_2("seed-two", "Hallo"))
print(task1_2("seed-three", "Hullo"))
    
```

```python
#Task 2
# Task 2.1
import random

def task2_1(filename, quantity, maximum):

    file = open(filename, "w")

    for i in range(quantity):
        num = random.randint(0, maximum)
        file.write(str(num) + "\n")

    file.close()

# Task 2.2
def task2_2(list_of_integers):

    if len(list_of_integers) <= 1:
        return list_of_integers

    mid = len(list_of_integers) // 2

    left = task2_2(list_of_integers[:mid])
    right = task2_2(list_of_integers[mid:])

    result = []

    while len(left) > 0 and len(right) > 0:

        if left[0] <= right[0]:
            result.append(left.pop(0))
        else:
            result.append(right.pop(0))

    result.extend(left)
    result.extend(right)

    return result

# Task 2.3

def task2_3(filename_in, filename_out):

    file = open(filename_in, "r")

    numbers = []

    for line in file:
        numbers.append(int(line.strip()))

    file.close()

    sorted_numbers = task2_2(numbers)

    file = open(filename_out, "w")

    for number in sorted_numbers:
        file.write(str(number) + "\n")

    file.close()
```

```python
# Task 3.1
def task3_1(quantity_of_data):
    units = {
        "KB": 10**3,
        "MB": 10**6,
        "GB": 10**9,
        "TB": 10**12
    }

    for unit in units:
        if quantity_of_data.endswith(unit):
            num = quantity_of_data[:-len(unit)]

            if num.isdigit():
                ans = int(num) * units[unit]
                print(ans)
                return ans

    print("invalid data")
    return "invalid data"


#main
task3_1("8KB")      
task3_1("12MB")     
task3_1("2GB")      
task3_1("100TB")      
task3_1("67B")   

#Task 3.2
def task3_2(quantity_of_data):
    units = {
        "KB": 10**3,
        "KiB": 2**10,
        "MB": 10**6,
        "MiB": 2**20,
        "GB": 10**9,
        "GiB": 2**30,
        "TB": 10**12,
        "TiB": 2**40
    }

    for unit in units:
        if quantity_of_data.endswith(unit):
            num = quantity_of_data[:-len(unit)]

            if num.isdigit():
                ans = int(num) * units[unit]
                print(ans)
                return ans

    print("invalid data")
    return "invalid data"


#main
task3_2("8KB")
task3_2("2MiB")      
task3_2("5GiB")
task3_2("1TiB")
task3_2("10ABC")

#Task 3.3
def task3_3(quantity_of_data, target_unit):

    units = {
        "KB": 10**3,
        "KiB": 2**10,
        "MB": 10**6,
        "MiB": 2**20,
        "GB": 10**9,
        "GiB": 2**30,
        "TB": 10**12,
        "TiB": 2**40
    }

    if target_unit not in units:
        print("invalid data")
        return "invalid data"

    bytes_value = task3_2(quantity_of_data)

    if bytes_value == "invalid data":
        print("invalid data")
        return "invalid data"

    ans = bytes_value / units[target_unit]
    print(ans)
    return ans


#main
task3_3("512MiB", "GiB")   
task3_3("1TB", "GB")       
task3_3("2048KiB", "MiB")  
```
